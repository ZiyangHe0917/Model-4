```python
import pandas as pd

df = pd.read_csv("allbacks.csv")
df.head()

books = pd.read_csv("allbacks.csv")

books
```

```python
sorted_covers = df['cover'].sort_values().reset_index(drop=True)
median_index = len(sorted_covers) // 2
print("Median Cover Type:", sorted_covers[median_index])
```

```python
mean_weight = books["weight"].mean()

print("Mean weight of all books:", mean_weight, "grams")
```

```python
mean_by_cover = books.groupby("cover")["weight"].mean()

print(mean_by_cover)
```

### Interpretation of Group Means

Hardback books have an average weight of 796.43 grams,
while paperback books have an average weight of 628.13 grams.
The difference is approximately 168.30 grams, suggesting
that hardback books in this sample tend to be heavier
than paperback books.
